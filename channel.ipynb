```python
import pandas as pd
df = pd.read_csv("TV_channel_dataset.csv")
df.head()
```

```python
df["Channel_Name"] = df["Channel_Name"].str.strip()
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df.loc[(df["Avg_Rating"] < 0) | (df["Avg_Rating"] > 5), "Avg_Rating"] = None
```

```python
df.loc[df["Subscription_Fee"] < 0, "Subscription_Fee"] = None
```

```python
for col in df.columns:
    if df[col].dtype == "object":
        converted = pd.to_numeric(df[col], errors="coerce")
        if converted.notna().sum() == df[col].notna().sum():
            df[col] = converted
```

```python
df = df.replace(["NA", "N/A", "unknown", "Unknown", ""], pd.NA)
print(df.isnull().sum())
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate records:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```
